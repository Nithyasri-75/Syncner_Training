```
from pyspark.sql.functions import avg, max, col
```

```
df = spark.read.option("header", True).option("inferSchema", True).csv("/FileStore/tables/ipl_over_data-2.csv")
df.display()
```

```
team_scores = df.select("year", col("team1_name").alias("team"), col("team1_score").cast("int").alias("score")) \
                        .union(df.select("year", col("team2_name").alias("team"), col("team2_score").cast("int").alias("score")))
team_scores.display()
```

```
avg_score_pivot = team_scores.groupBy("team").pivot("year").agg(avg("score").alias("avg_score"))
avg_score_pivot.display()
```

```
max_score_pivot = team_scores.groupBy("team").pivot("year").agg(max("score").alias("max_score"))
max_score_pivot.display()
```
