```
import pyspark
from pyspark.sql.functions import col, to_timestamp, to_date, date_format
```

```
df=spark.read.format('csv').option('inferSchema',True).option('header',True).load('/FileStore/tables/ipl_over_data.csv')
df.display()
```

```
df = df.withColumn("datetime", to_timestamp(col("match_datetime"))) \
       .withColumn("match_date", to_date(col("match_datetime"))) \
       .withColumn("match_time", date_format(col("match_datetime"), "HH:mm:ss"))

df.select( "datetime", "match_date", "match_time").display()
```
