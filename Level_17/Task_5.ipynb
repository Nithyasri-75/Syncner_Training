## 1. Number of Teams Each Year

```
from pyspark.sql import functions as F
from pyspark.sql.functions import col,array,explode,trim,avg,countDistinct,count,collect_list, collect_set
```

```
df = spark.read.option("header", True).option("inferSchema", True).csv("/FileStore/tables/ipl_commentary_data-4.csv")
df.display()
```

```
teams_by_year = df.select("year", trim("team1_name").alias("team1"), trim("team2_name").alias("team2")) \
    .selectExpr("year", "team1", "team2") \
    .withColumn("team", explode(array("team1", "team2"))) \
    .drop("team1", "team2") \
    .dropna(subset=["team"]) \
    .distinct() \
    .groupBy("year") \
    .count() \
    .withColumnRenamed("count", "number_of_teams")

display(teams_by_year)
```

## 3. Average number of ball_no per match per inning

```
df2 = spark.read.option("header", True).option("inferSchema", True).csv("/FileStore/tables/export.csv")
df2.display()
```

```
balls_per_inning = df2.groupBy("match_id", "inning").agg(count("ball_no").alias("balls_in_inning"))
avg_balls = balls_per_inning.groupBy("inning").agg(avg("balls_in_inning").alias("avg_balls_per_match_per_inning"))
avg_balls.orderBy("inning").display()
```

### 4. Number of matches each team has won in each year, comparing each teams performance across the years. Sort accordingly

```
df.select("match_winning_team").distinct().display()
```

```
df = df.filter(col("match_winning_team").isNotNull())
df.display()
```

```
team_wins_per_year = df.filter(col("match_winning_team").isNotNull()) \
    .select("year", "match_no", "match_winning_team") \
    .dropDuplicates() \
    .groupBy("year", "match_winning_team") \
    .agg(count("*").alias("matches_won"))
sorted_wins = team_wins_per_year.orderBy("year", col("matches_won").desc())
sorted_wins.display()
```

### 5. Compare collect_list and collect_set to get the list of batsmen in the match for each team using the above extracted column

```
from pyspark.sql import SparkSession
from pyspark.sql.functions import collect_list, collect_set
```

```
df3 = spark.read.option("header", True).option("inferSchema", True).csv("/FileStore/tables/ipl_over_data-1.csv")
df3.display()
```

collect_list

```
batsmens = df3.groupBy("team1_name", "team2_name")\
            .agg(collect_list("over_batsman1_name").alias("team1_batsmen_all"),
                collect_list("over_batsman2_name").alias("team2_batsmen_all"))
batsmens.display()
```

collect_set

```
batsmen = df3.groupBy("team1_name", "team2_name").agg(collect_set("over_batsman1_name").alias("team1_batsmen_unique"),collect_set("over_batsman2_name").alias("team2_batsmen_unique"))
batsmen.display()
```
