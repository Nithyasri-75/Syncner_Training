```
import pyspark
```

```
matches_df = spark.read.option("header", True).option("inferSchema", True).csv("/FileStore/tables/ipl_commentary_data-2.csv")
matches_df.display()
```

```
tied_matches_df = matches_df.filter(matches_df["match_status"] == "tied")
tied_matches_df.display()
```

```
tied_matches_df.select("match_id", "year", "team1_name", "team2_name", "match_tie_breaker").display()
```

```
tiebreakers_df = tied_matches_df.select("match_tie_breaker").distinct()
tiebreakers_df.display()
```
